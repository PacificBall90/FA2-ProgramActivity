```python
num1 = int(input("Enter number: "))
num2 = int(input("Enter another number: "))
num3 = int(input("Enter another number: "))

greatest_num = 0

print()
print("Comparing...")
print()

if num1 > greatest_num:
  greatest_num = num1
if num2 > greatest_num:
  greatest_num = num2
if num3 > greatest_num:
  greatest_num = num3

print("The greatest number is:", greatest_num)
```
